```python
from youtube_transcript_api import YouTubeTranscriptApi ,TranscriptsDisabled
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_nvidia_ai_endpoints import ChatNVIDIA
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
```

```python
video_id = "FrRjf9muIyY"
try:
  api = YouTubeTranscriptApi()
  transcript_list = api.fetch(video_id,languages=["en"])
  transcript = " ".join(chunk.text for chunk in transcript_list)
  print(transcript)

except TranscriptsDisabled:
  print("transcript are not availabe")
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap =200)
chunks = splitter.create_documents([transcript])
```

```python
chunks[5]
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="BAAI/bge-m3"
)
vector_store = FAISS.from_documents(chunks, embeddings)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(["6a08bff7-e466-4985-855b-f6aa7273c81e"])
```

```python
retriever = vector_store.as_retriever(search_type="similarity" ,search_kwargs={"k":4})
retriever
```

```python
retriever.invoke("how to apply in google")
```

```python
llm = ChatNVIDIA(
    model="nvidia/nemotron-3-nano-30b-a3b"
)
```

```python
prompt = PromptTemplate(
    template ="""You are helpfull assistent.
    Answer only from the Provided transcript context.
    if context is insufficient just say i don't know.
    {context}
    Question : {question}
    """,
    input_variables = ['context','question']
)
```

```python
question = "does dsa related tips are there in video"
retrived_docs = retriever.invoke(question)
```

```python
retrived_docs
```

```python
context_text = "\n\n".join(doc.page_content for doc in retrived_docs)
context_text
```

```python
final_prompt = prompt.invoke({"context":context_text ,"question":question})
final_prompt
```

```python
answer = llm.invoke(final_prompt)
print(answer.content)
```
